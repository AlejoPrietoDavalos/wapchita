```python
from typing  import Type, Generic

from requests import Response
from requestsdantic import T_BaseResponse

class BaseResponseNew(Generic[T_BaseResponse]):
    def __init__(self, *, response: Response, cls_response: Type[T_BaseResponse]):
        self._response = response
        self.cls_response = cls_response

    @property
    def response(self) -> Response:
        return self._response

    @property
    def status_code(self) -> int:
        return self.response.status_code

    @property
    def response_data(self) -> T_BaseResponse:
        return self.cls_response(data=self.response.json())


class ResponseGroupsNew(BaseResponseNew[ResponseGroups]):
    def __init__(self, *, response: Response):
        super().__init__(response=response, cls_response=ResponseGroups)

asd = ResponseGroupsNew(response=response)
```
